# Data Cleaning

```python
df=df.drop_duplicates().reset_index(drop=True)
```

```python
df.columns=df.columns.str.lower()
```

```python
df.drop("legacy_system_id",axis=1,inplace=True)
```

 Dropping legacy_system_id ~96% missing , not usable as a feature

```python
for col in df.select_dtypes(include="object"):
    print(col)
    print(df[col].unique())
```

```python
df["customer_name"]=df["customer_name"].str.title()
```

```python
country_map={'usa':'United States',
            'USA':'United States'}

df["country"]=df["country"].replace(country_map)
```

```python
industry_map={'information technology':'IT',
              'Information Technology':'IT'}

df["industry"]=df["industry"].replace(industry_map)
```

Normalizing inconsistent casing found in country and industry column 

```python
df["payment_method"]=df["payment_method"].str.title()
```

```python
df["satisfaction_rating"]=df.groupby("subscription_plan")["satisfaction_rating"].transform(lambda x:x.fillna(x.median()))
```

Imputing missing satisfaction_rating with the median per subscription_plan,since satisfaction likely varies systematically by plan tier

```python
df["payment_method"] = df.groupby("billing_cycle")["payment_method"].transform(lambda x:x.fillna(x.mode()[0]))
```

Imputing missing payment_method with the most common method per billing_cycle

```python
df[df["avg_response_time_hours"].isna()]["support_tickets"].value_counts()
```

avg_response_time_hours is NaN only for customers with zero support tickets (confirmed above) — these customers have no response time to record, so we
leave the value as NaN rather than filling with 0 or somethimg else. Filling it would incorrectly suggest an instant response time, which would distort any avg/correlation involving this column.

# Feature Engineering

```python
bins = [0, 12, 24, 48, df["customer_tenure_months"].max()+1]

labels = ["New Customer","Developing Customer","Established Customer","Loyal Customer"]


df["customer_lifecycle"] = pd.cut(
    df["customer_tenure_months"],
    bins=bins,
    labels=labels,
    right=False
)

bins = [0, 8, 16, 24, df["last_month_login_count"].max()+1]

labels = ["Rare Visitor","Occasional User","Consistent User","Daily User"]


df["customer_engagement"] = pd.cut(
    df["last_month_login_count"],
    bins=bins,
    labels=labels,
    right=False
)

bins = [0, 1, 6, 10, df["support_tickets"].max()+1]

labels = ["No Support Needed","Occasional Assistance","Frequent Assistance","Ongoing Support"]


df["support_experience"] = pd.cut(
    df["support_tickets"],
    bins=bins,
    labels=labels,
    right=False
)


bins = [18, 26, 36, 46, 56, df["age"].max()+1]

labels = [
    "18-26",
    "27-36",
    "37-46",
    "47-56",
    "57+"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels,
    right=False
)

df["revenue_segment"] = pd.qcut(
    df["total_revenue"],
    q=4,
    labels=[
        "Starter Account",
        "Growth Account",
        "Business Account",
        "Enterprise Account"
    ]
)

def retention_priority(row):

    if (
        row["last_month_login_count"] < 10 and
        row["support_tickets"] >= 6 and
        row["satisfaction_rating"] <= 2
    ):
        return "Immediate Outreach"

    elif (
        row["last_month_login_count"] < 18 or
        row["support_tickets"] >= 3 or
        row["satisfaction_rating"] == 3
    ):
        return "Monitor Engagement"

    else:
        return "Loyal Customer"


df["retention_priority"] = df.apply(retention_priority, axis=1)

```

Creating derived categorical features (lifecycle stage, engagement level, support experience, age group, revenue segment, retention priority) from cleaned numeric columns to support segmentation and dashboard analysis

```python
df.to_csv("saas_churn_analysis_clean.csv",index=False)
```
